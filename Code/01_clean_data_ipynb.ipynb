```python
import pandas as pd

# ---------------- ตั้งค่า ----------------
INPUT_FILE = "c:/Users/User/Desktop/68030880/My project/Salary_Data 7.csv"          # แก้ path ตรงนี้ได้
OUTPUT_FILE = "Salary_Data_cleaned.csv"
DROP_DUPLICATES = True                    # False = ไม่ลบแถวซ้ำ
MIN_SALARY = 10_000                       # เงินเดือนต่ำกว่านี้ถือว่าผิดปกติ

# ---------------- โหลดข้อมูล ----------------
df = pd.read_csv(INPUT_FILE)
n_raw = len(df)

# ---------------- จัดรูปแบบข้อความ ----------------
for col in ["Gender", "Education Level", "Job Title"]:
    df[col] = df[col].str.strip().str.replace(r"\s+", " ", regex=True)

df["Education Level"] = df["Education Level"].replace({
    "Bachelor's": "Bachelor's Degree",
    "Master's": "Master's Degree",
    "phD": "PhD",
})

df["Job Title"] = df["Job Title"].replace({
    "Juniour HR Coordinator": "Junior HR Coordinator",
    "Juniour HR Generalist": "Junior HR Generalist",
    "Social M": "Social Media Manager",
    "Social Media Man": "Social Media Manager",
    "Front end Developer": "Front End Developer",
    "Back end Developer": "Back End Developer",
    "Customer Service Rep": "Customer Service Representative",
})

# ---------------- ลบค่าว่าง ----------------
df = df.dropna()

# ---------------- ลบค่าผิดปกติ ----------------
df = df[df["Salary"] >= MIN_SALARY]                          # เงินเดือนต่ำเกินจริง
df = df[df["Years of Experience"] <= df["Age"] - 16]         # ประสบการณ์เกินอายุ

# ---------------- ลบแถวซ้ำ ----------------
if DROP_DUPLICATES:
    df = df.drop_duplicates()

# ---------------- ปรับชนิดข้อมูลและบันทึก ----------------
df = df.astype({"Age": int, "Salary": int}).reset_index(drop=True)
df.to_csv(OUTPUT_FILE, index=False)

print(f"ก่อนคลีน {n_raw:,} แถว -> หลังคลีน {len(df):,} แถว")
print(f"บันทึกไฟล์: {OUTPUT_FILE}")
```
